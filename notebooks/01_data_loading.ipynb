```python
import pandas as pd
import numpy as np

df = pd.read_parquet('../data/raw/yellow_tripdata_2024-01.parquet')
zones = pd.read_csv('../data/raw/taxi_zone_lookup.csv')

print("=" * 55)
print("  NYC YELLOW TAXI - JANUARY 2024")
print("  UrbanMobility 360 - Data Loading")
print("=" * 55)

print(f"\nTrip Data:")
print(f"  Rows:    {df.shape[0]:>12,}")
print(f"  Columns: {df.shape[1]:>12}")

print(f"\nColumn Names & Types:")
for col, dtype in df.dtypes.items():
    print(f"  {col:30s} {str(dtype)}")

print(f"\nZone Lookup:")
print(f"  Total zones: {len(zones)}")
print(f"  Boroughs:    {zones['Borough'].nunique()}")
print(zones['Borough'].value_counts())

print(f"\nMissing Values:")
missing = df.isnull().sum()
missing = missing[missing > 0]
if len(missing) > 0:
    for col, count in missing.items():
        pct = count / len(df) * 100
        print(f"  {col:30s} {count:>10,} ({pct:.1f}%)")
else:
    print("  No missing values!")

print(f"\nQuick Stats:")
print(f"  Avg distance:  {df['trip_distance'].mean():.2f} miles")
print(f"  Avg fare:      ${df['fare_amount'].mean():.2f}")
print(f"  Min datetime:  {df['tpep_pickup_datetime'].min()}")
print(f"  Max datetime:  {df['tpep_pickup_datetime'].max()}")
```

```python
# Check zone matching
trip_zone_ids = set(df['PULocationID'].unique()) | set(df['DOLocationID'].unique())
lookup_zone_ids = set(zones['LocationID'].unique())
matched = trip_zone_ids & lookup_zone_ids
unmatched = trip_zone_ids - lookup_zone_ids

print(f"Zone IDs in trip data:  {len(trip_zone_ids)}")
print(f"Zone IDs in lookup:     {len(lookup_zone_ids)}")
print(f"Matched:                {len(matched)}")
print(f"Unmatched:              {len(unmatched)}")

if len(unmatched) > 0:
    print(f"\nUnmatched IDs (will be removed during cleaning):")
    print(f"  {sorted(unmatched)}")
```
